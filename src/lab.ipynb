```python
import os
import shutil

def organize_files(csv_dir, png_dir, output_dir):
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Get list of CSV and PNG files
    csv_files = [f for f in os.listdir(csv_dir) if f.endswith('.csv')]
    png_files = [f for f in os.listdir(png_dir) if f.endswith('.png')]

    # Debugging: Print the lists of files
    print("CSV Files:", csv_files)
    print("PNG Files:", png_files)

    # First, move all CSV files to new directories
    for csv_file in csv_files:
        base_name = os.path.splitext(csv_file)[0]
        new_dir = os.path.join(output_dir, base_name)
        os.makedirs(new_dir, exist_ok=True)

        # Move CSV file to the new directory
        shutil.move(os.path.join(csv_dir, csv_file), os.path.join(new_dir, csv_file))

        # Debugging: Print the new directory and moved file
        print(f"Created directory: {new_dir}")
        print(f"Moved {csv_file} to {new_dir}")

    # Now, move all matching PNG files to their respective directories
    for png_file in png_files:
        png_base_name = os.path.splitext(png_file)[0]
        for csv_file in csv_files:
            csv_base_name = os.path.splitext(csv_file)[0]
            if csv_base_name in png_base_name:
                new_dir = os.path.join(output_dir, csv_base_name)
                shutil.move(os.path.join(png_dir, png_file), os.path.join(new_dir, png_file))

                # Debugging: Print the new directory and moved file
                print(f"Moved {png_file} to {new_dir}")


# Define your directories
csv_directory = "E:\Economic_Data\Output data"
png_directory = "E:\Economic_Data\Output data\output"
output_directory = "E:\Economic_Data\Output data\Economic Data (Investing.com)  Analysis"

# Run the function
organize_files(csv_directory, png_directory, output_directory)
```

```python
import pandas as pd
import glob

# Assuming all your CSV files are in the same directory, e.g., "data/"
csv_files = glob.glob("E:\SignalModel\minute price\*.csv")

# Read and concatenate all CSV files
dataframes = []
for file in csv_files:
    df = pd.read_csv(file, parse_dates=['datetime'])
    dataframes.append(df)

# Concatenate all DataFrames
merged_df = pd.concat(dataframes)

# Remove duplicate datetime entries
merged_df = merged_df.drop_duplicates(subset='datetime')

# Sort by datetime if necessary
merged_df = merged_df.sort_values(by='datetime')

# Save to a new CSV file
merged_df.to_csv("merged_dataset.csv", index=False)
```
