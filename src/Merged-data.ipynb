```python
import pandas as pd
import glob
import os
import re
```

```python
# Directory where the datasets are stored
directory = r'E:\Economic_Data\Input data\\'

# Get a list of all CSV files in the directory
file_paths = glob.glob(os.path.join(directory + '*.csv'))
```

```python
# Check if any files were found
if not file_paths:
    print("No CSV files found in the directory.")
else:
    # Initialize an empty list to store dataframes
    dataframes = []

    # Process each file
    for file_path in file_paths:
        try:
            # Load the dataset
            df = pd.read_csv(file_path)
            
            # Fill missing 'Date' values with the last non-missing value
            df['Date'] = df['Date'].fillna(method='ffill')
            
            # Combine 'Date' and 'Time' columns into a single 'datetime' column
            df['datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'])
            
            # Drop the original 'Date' and 'Time' columns
            df = df.drop(columns=['Date', 'Time'])
            
            # Append the processed dataframe to the list
            dataframes.append(df)
        
        except Exception as e:
            print(f"Error processing file {file_path}: {e}")

    # Check if any dataframes were successfully processed
    if not dataframes:
        print("No valid dataframes to concatenate.")
    else:
        # Concatenate all dataframes into a single dataframe
        merged_df = pd.concat(dataframes, ignore_index=True)
        
        # Sort the dataframe by 'datetime' column in ascending order
        merged_df = merged_df.sort_values(by='datetime')
        
        # Save the combined dataframe to a CSV file
        output_file = os.path.join(directory, 'merged_datasets.csv')
        merged_df.to_csv(output_file, index=False)
        
        print(f"Datasets have been successfully merged and saved as '{output_file}'.")
```

```python
directory2= r'E:\Economic_Data\Output data\\'
        
# Extract the base event name by removing the month part in parentheses
merged_df['base_event'] = merged_df['Event'].apply(lambda x: re.split(r' \(', x)[0])

# Identify unique base event names
unique_events = merged_df['base_event'].unique()

# Create separate dataframes for each unique base event and save as CSV
for event in unique_events:
    event_df = merged_df[merged_df['base_event'] == event]
    event_file = os.path.join(directory2, f"{event.replace('/', '_').replace(' ', '_')}.csv")
    event_df.drop(columns=['base_event'], inplace=True)
    event_df.to_csv(event_file, index=False)
    print(f"Saved {event} dataset as '{event_file}'")
```
